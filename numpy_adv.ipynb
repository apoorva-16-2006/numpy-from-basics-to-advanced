# Advanced Numpy

✅ What is Advanced Indexing?

Advanced indexing means:

👉 Indexing an array using lists, arrays, boolean masks, or multiple indices to select specific elements.

It is more powerful than normal slicing.

✅ Types of Advanced Indexing
✅ 1. Integer Array Indexing

You use a list/array of indices to pick specific elements.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
idx = [0, 2, 4]

print(arr[idx])
```

```python
arr[[1,4]]
```

✅ 2. Integer Indexing on 2D Array

```python
arr = np.array([
    [10, 11, 12],
    [20, 21, 22],
    [30, 31, 32]
])

rows = [0, 2]
cols = [1, 2]

print(arr[rows, cols])
```

Meaning:

(row 0, col 1) → 11

(row 2, col 2) → 32

👉 Picks pair-wise elements.

✅ 3. Boolean Indexing (VERY IMPORTANT)

Boolean indexing means:

👉 Selecting elements from an array using True/False conditions.

If condition is True → element is selected
If condition is False → element is skipped

```python
arr = np.array([5, 12, 18, 7, 3])
print(arr[arr > 10]) #Meaning → Pick values greater than 10.
```

✅ Boolean Indexing in 2D

```python
arr = np.array([
    [10, 50],
    [30, 5]
])

print(arr[arr > 20])
```

```python
names = np.array(["Ram", "Amit", "Ram", "Sita"])
marks = np.array([85, 90, 75, 95])
print(marks[names == "Ram"])
```

✅ Why do we create arrays like this?

Because now each index matches:

Index	Name	Marks
0	Ram	85
1	Amit	90
2	Ram	75
3	Sita	95

That means:

names[0] → "Ram"

marks[0] → 85

✅ They belong together.

Step 1: Create a boolean mask

names == "Ram"

Output:

[ True False  True False ]

This means:

index 0 → Ram ✅

index 1 → Amit ❌

index 2 → Ram ✅

index 3 → Sita ❌

✅ Step 2: Use mask to filter marks

marks[names == "Ram"]

NumPy will select only the places where mask is True:

✅ For index 0 → 85
✅ For index 2 → 75

Final Output:

[85 75]

✅ How it works in simple language

Check which names are "Ram".

Wherever it is True, pick the marks from the marks array.

Return only those marks.

✅ What is Broadcasting? (Easy Explanation)

Broadcasting = NumPy automatically expands smaller arrays so they can match bigger arrays during operations.

👉 It allows you to do math between arrays of different shapes.

✅ You do NOT need loops

NumPy does it automatically.

✅ Very simple rule
✅ Broadcasting works when:

Dimensions are equal, OR

One of them is 1 (NumPy stretches it), OR

One array has missing dimensions (NumPy adds automatically)

```python
arr=np.array([20,30,40,50])
print(arr+5) # 5 is add with all the elements in arr 
#Why?
#5 becomes [5, 5, 5,5] automatically (broadcasting).
```

```python
arr_2=np.array([[1],[2],[3],[4]])
print(arr+arr_2)
```

```python
a = np.array([1, 2, 3])
b = np.array([[10],
              [20],
              [30]])
print(a + b)
```

✅ How Broadcasting works here

a shape → (3,)
b shape → (3,1)

NumPy expands them:

a → [1 2 3]

      ↓ ↓ ↓ expands down

      ↓ ↓ ↓
b → [[10 10 10]

     [20 20 20]

     [30 30 30]]

Then adds corresponding values.

✅ Rule of Broadcasting

✅ Broadcasting works only if:

Arrays have same number of dimensions (NumPy adds 1s if needed)

In every dimension:
sizes must match OR one of them must be 1

❌ If a dimension is different and not 1, broadcasting fails.

✅ Allowed when:

numbers are equal

OR one of them is 1

```python
arr=np.arange(1,10).reshape(3,3)
row=np.array([10,20,30])
print(arr+row)
```

Why?

Row → [10,20,30] broadcasts to all 3 rows.

```python
col=np.array([[10],
              [20],
              [30]]) #Column broadcasts to each column.
print(row+col)
```

❌ Example 1 — Broadcasting Not Possible

Arrays:

A = np.array([1, 2, 3])        # Shape → (3,)

B = np.array([10, 20])         # Shape → (2,)

✅ Check broadcasting rules

Compare shapes from the end:

A : (3,)
B : (2,)

3 ≠ 2

No dimension is 1
👉 Broadcasting fails

❌ Example 2 — 2D & 2D mismatch
Arrays:

A = np.array([[1,2,3],

              [4,5,6]])       # Shape → (2, 3)

B = np.array([[10,20],

              [30,40]])       # Shape → (2, 2)

Compare shapes (from the end):
A : (2, 3)
B : (2, 2)

Check each dimension:

A	B	Valid?

3	2	❌ No match, none is 1

2	2	✅ match

Because 3 ≠ 2 → no stretching possible.

```python
A = np.zeros((3, 4, 5))     # Shape → (3, 4, 5)
B = np.ones((4, 1))         # Shape → (4, 1)
print(A+B)
```

## Working With Mathematical Formulas

```python
# 1/1+e**-x
def sigmoid_function(array):
    return (1/(1+np.exp(-(array))))

a=np.arange(100)
sigmoid_function(a)
```

```python
actual=np.random.randint(1,50,25)
predicted=np.random.randint(1,50,25)
print(actual)
print(predicted)
```

```python
def mse(actual,predicted):
    return np.mean((actual-predicted)**2)

print(mse(actual,predicted))
```

### Find missing values

```python
import numpy as np

arr = np.array([10, 20, np.nan, 40])
result = np.isnan(arr)

print(result)
```

```python
np.sum(np.isnan(arr)) #Count missing values
```

```python
np.any(np.isnan(arr)) #Check if any missing value exists it returns True 
```

```python
np.where(np.isnan(arr)) #Find index of missing value
```

```python
arr2 = np.array([[1, 2], [np.nan, 4]])

np.isnan(arr2) #2D Array Example
```

```python
arr3=np.array([[[1,2],[3,4]],[[5,np.nan],[7,8]]])
np.isnan(arr3) #3D Array Example
```
